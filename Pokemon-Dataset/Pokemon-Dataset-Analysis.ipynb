Which Pokémon has the highest Total Stats?

```python
# Creating the total_stats column
data['total_stats']=data['HP']+data['Attack']+data['Defense']+data['Sp. Atk']+data['Sp. Def']+data['Speed']
data
```

```python
data.loc[data['total_stats'].idxmax()]
```

How many Legendary and Non-Legendary Pokémon are there?

```python
data['Legendary'].value_counts()
```

Do Legendary Pokémon have higher average Total Stats?

```python
data.groupby('Legendary')['total_stats'].mean()
```

Which Pokémon Type (Type 1) has the highest average Total Stats?

```python
data.groupby('Type 1')['total_stats'].mean().sort_values(ascending=False)
```

What is the distribution of Pokémon generations?

```python
data['Generation'].value_counts().sort_index()
```

Which stat has the highest variability?

```python
data[['HP','Attack','Defense','Sp. Atk','Sp. Def','Speed']].std()
```

Is Attack correlated with Speed?

```python
data[['Attack','Speed']].corr()
```

Are single-type Pokémon more common than dual-type Pokémon?

```python
data['Type 2'].isnull().value_counts()
```
